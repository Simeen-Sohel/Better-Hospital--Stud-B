```python
import pandas as pd
```

```python
df = pd.read_csv('/content/Bed_Occupancy_Data.csv')
print("First 5 rows of the dataset:")
display(df.head())
```

```python
print("\nMissing values in each column:")
display(df.isnull().sum())
```

```python
# Identify numerical columns for outlier detection
numerical_cols = df.select_dtypes(include=['number']).columns

print("Outlier Detection Report (using IQR method):")
print("-------------------------------------------")

outlier_counts = {}

for column in numerical_cols:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Count outliers
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    num_outliers = len(outliers)
    outlier_counts[column] = num_outliers

    print(f"Column '{column}': {num_outliers} outliers found.")

print("\nSummary of Outliers:")
display(pd.DataFrame(outlier_counts.items(), columns=['Column', 'Number of Outliers']))
```

```python
# Ensure TRANS_EndDate and TRANS_EndTime are in datetime format
df['TRANS_EndDate'] = pd.to_datetime(df['TRANS_EndDate'], format='%d/%m/%Y', errors='coerce')
df['TRANS_EndTime'] = pd.to_datetime(df['TRANS_EndTime'], format='%H:%M:%S', errors='coerce')

# Drop rows where TRANS_EndDate or TRANS_EndTime are null
original_rows = df.shape[0]
df_cleaned = df.dropna(subset=['TRANS_EndDate', 'TRANS_EndTime']).copy()

print(f"Original number of rows: {original_rows}")
print(f"Number of rows after dropping missing values: {df_cleaned.shape[0]}")
print("Missing values in TRANS_EndDate after cleaning:", df_cleaned['TRANS_EndDate'].isnull().sum())
print("Missing values in TRANS_EndTime after cleaning:", df_cleaned['TRANS_EndTime'].isnull().sum())
```

```python
# Save the cleaned DataFrame to a new CSV file
output_filename = 'Bed_Occupancy_Data_cleaned.csv'
df_cleaned.to_csv(output_filename, index=False)

print(f"Cleaned data saved to {output_filename}")
```
